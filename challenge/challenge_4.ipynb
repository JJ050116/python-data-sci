## Debug checkpoint
Pandas ใช้ `&` สำหรับเชื่อม 2 เงื่อนไข และต้องมีวงเล็บรอบแต่ละเงื่อนไข

```python
import pandas as pd
df = pd.read_csv("../data/footballers.csv")

# ลองรัน: Error นี้บอกอะไรเกี่ยวกับ and และ Series?
df[(df["position"].isin(["ST", "RW"])) and (df["physical"] >= 75)]
```

### เขียนคำตอบของคุณด้านล่าง
เขียน 1 ประโยคว่าเหตุใดผลลัพธ์จึงควรมีเฉพาะบางแถว ไม่ใช่ทั้งตาราง

```python
# ==============================================================================
# วิธีที่ 1: ใช้ Boolean Masking ด้วย & (วิธีมาตรฐานของ Pandas)
# ==============================================================================
# 1.1 สร้าง Boolean Series (True/False) สำหรับเช็กตำแหน่งว่าเป็น ST หรือ RW
is_forward = df["position"].isin(["ST", "RW"])

# 1.2 สร้าง Boolean Series สำหรับเช็กว่า physical >= 75
is_strong = df["physical"] >= 75

# 1.3 เชื่อมเงื่อนไขด้วย & (Bitwise AND) และครอบวงเล็บ () รอบแต่ละเงื่อนไข
# - ทำไมต้องใช้ & : เพราะ & เทียบ True/False ทีละแถว (Element-wise) ต่างจาก 'and' ที่ใช้กับค่าเดี่ยว
# - ทำไมต้องมี () : เพราะ & มีลำดับความสำคัญสูงกว่า >= หรือ == จึงต้องใส่ () เพื่อคำนวณเงื่อนไขย่อยก่อน
# - [["name", "position", "physical"]] : ดึงเฉพาะคอลัมน์ที่ต้องการแสดงผล
strong_forwards_1 = df[is_forward & is_strong][["name", "position", "physical"]]
display(strong_forwards_1)

# ==============================================================================
# วิธีที่ 2: ใช้คำสั่ง query() (ทางเลือกที่เขียนสั้น อ่านง่าย)
# ==============================================================================
# - ทำไมถึงเขียนแบบนี้ : query() รับเงื่อนไขเป็น String ทำให้สามารถใช้คำว่า 'and', 'or', 'in' ได้เลย
# - ไม่ต้องใส่เครื่องหมาย & และไม่ต้องครอบวงเล็บ () ในแต่ละเงื่อนไข
strong_forwards_2 = df.query("position in ['ST', 'RW'] and physical >= 75")[["name", "position", "physical"]]
display(strong_forwards_2)
```

```python
# 💡 คำอธิบายจุดที่เคยติด Error ก่อนหน้านี้:
# การใช้ df.query("physical >= 75") จะได้ผลลัพธ์เป็น DataFrame ไม่ใช่ Boolean Series
# เมื่อนำ attack (Series) ไป & กับ strong (DataFrame) จึงเกิด TypeError
# การแก้ให้ถูกต้องคือต้องเปลี่ยน strong ให้เป็น Boolean Series ดังนี้:
attack = df["position"].isin(["ST", "RW"])
strong = df["physical"] >= 75  # เป็น Boolean Series (True/False)

strong_forwards = df[attack & strong][["name", "position", "physical"]]
display(strong_forwards)
```
